# Customer Segmentation Project
K-Means clustering using purchase behavior and demographic variables.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv('../data/customer_data.csv')
df.head()
```

```python
features = ['Annual_Income_INR', 'Purchase_Frequency', 'Average_Order_Value_INR', 'Annual_Spend_INR', 'Recency_Days', 'Online_Purchase_Ratio', 'Discount_Usage_Ratio', 'Loyalty_Score']
X = StandardScaler().fit_transform(df[features])

scores={}
for k in range(3,7):
    labels=KMeans(n_clusters=k,random_state=42,n_init=20).fit_predict(X)
    scores[k]=silhouette_score(X,labels)
scores
```

**Selected number of clusters:** 3 (highest silhouette score among k=3..6).

```python
df.groupby('Segment').agg(Customers=('Customer_ID','count'), Avg_Spend=('Annual_Spend_INR','mean'), Frequency=('Purchase_Frequency','mean'), Recency=('Recency_Days','mean'), Loyalty=('Loyalty_Score','mean')).round(2)
```

```python
df.groupby('Segment')['Annual_Spend_INR'].mean().sort_values().plot(kind='barh', title='Average Annual Spend by Segment')
plt.xlabel('INR')
plt.tight_layout()
plt.show()
```
